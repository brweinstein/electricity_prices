# Testing Notebook

```python
import pandas as pd

url = "https://reports-public.ieso.ca/public/PriceNodal/PUB_PriceNodal_2024.csv"
df = pd.read_csv(url, skiprows=4)
df.head(10)
df["timestamp"] = pd.to_datetime(df["Date"]) + pd.to_timedelta(df["Hour"] - 1, unit="h")
toronto = df[["timestamp", "Darlington"]].rename(columns={"Darlington":"price_toronto"})
toronto.describe()
```

# Quick Graph

```python
import matplotlib.pyplot as plt

toronto["price_toronto"].clip(upper=200).hist(bins=100)
plt.show()
```

# By Day

```python
toronto["dow"] = toronto["timestamp"].dt.day_name()
toronto["hour"] = toronto["timestamp"].dt.hour

# median is more robust than mean given the spikes you just saw
toronto.groupby("dow")["price_toronto"].median().reindex(
    ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
)

toronto.groupby("hour")["price_toronto"].median()
```
